## Parte 1 — Baixando o HTML com `requests`


```python
import requests

URL = "https://quotes.toscrape.com/"

# Faz a requisição GET (pede o conteúdo da página)
res = requests.get(URL)

print("status_code:", res.status_code)          # 200 = deu certo
print("Caracteres no res.text:", len(res.text))  # tamanho do HTML

# Mostra apenas os primeiros 3000 caracteres do HTML cru
print("\n----- HTML CRU (início) -----")
print(res.text[:3000])
```

## Parte 2 — Extraindo os dados com `BeautifulSoup`


```python
from bs4 import BeautifulSoup

# Entrega o HTML (baixado com requests) ao BeautifulSoup
soup = BeautifulSoup(res.text, "lxml")

# find: primeiro elemento. Aqui, o título principal da página (<h1>)
titulo = soup.find("h1")
print("Título da página:", titulo.get_text(strip=True))
```

```python
# find_all: Encontra todas as caixas de citação
caixas = soup.find_all("div", class_="quote")
print("Citacoes encontradas:", len(caixas))
```

```python
# Para cada caixa, extraímos texto da citação + autor + tags
dados = []
for caixa in caixas:
    
    citacao = caixa.find("span", class_="text").get_text(strip=True)
    autor = caixa.find("small", class_="author").get_text(strip=True)
    
    tags = [tag.get_text(strip=True) for tag in caixa.find_all("a", class_="tag")]
    


    dados.append({"citacao": citacao,"autor": autor,"tags": tags,})

    print(f"Autor: {autor:20} | Citação: {citacao[:50]}... | Tags: {', '.join(tags)}")
```

### Tabela


```python
import pandas as pd

df = pd.DataFrame(dados)
print(df)

# Salvando em CSV (aparece no painel de arquivos do Colab)
df.to_csv("quotes.csv", index=False)
print("Arquivo quotes.csv salvo!")
```
